# Historic Data Validation with Orchestrator Function

This notebook validates historic DMS instances using the **orchestrator function** for parallel processing.

## How It Works:
1. **Select a view** to validate (from deployed workflows)
2. **Call orchestrator** - triggers parallel validation across N partitions
3. **Orchestrator manages retries** - partitions that timeout are automatically retried
4. **Monitor progress** - track partition execution
5. **View results** - check validation outcomes

## Key Features:
- ✅ **Parallel processing**: 10-20x faster than sequential
- ✅ **Runs in CDF**: No local processing needed
- ✅ **Fault tolerant**: Cursor-based resumption + automatic retries
- ✅ **Orchestrator manages everything**: Including partition retries
- ✅ **Scalable**: Handles millions of instances

## Important Notes:
- **Orchestrator can timeout**: After ~9 minutes, orchestrator saves state and exits
- **Resume orchestrator**: Just call it again with same config - it will resume monitoring
- **Partitions auto-retry**: Orchestrator retriggers incomplete partitions
- **Progress preserved**: All state saved in CDF (cursors, partition status)
