# Print Data Sample Function - Test & Deploy

This notebook helps you:
1. Test the function locally
2. Deploy to Cognite Functions
3. Call with sample data
4. View the logs to capture input structure


## Setup


```python

from cognite.client import CogniteClient
from thisisneat import NeatSession, get_cognite_client
import json

# Initialize client
client = get_cognite_client("get.env")
print(f"Connected to project: {client.config.project}")
```

## Deploy to Cognite Functions


```python
client.functions.delete(external_id="print-data-sample")
```

```python
# Deploy function
function = client.functions.create(
    name="print-data-sample",
    external_id="print-data-sample",
    folder=".",
    function_path="handler.py",
    description="Print data parameter to capture sample input structure",
    runtime="py311"
)

print(f"Function deployed: {function.external_id}")
print(f"Name: {function.name}")
print(f"ID: {function.id}")
```

```python
client.functions.retrieve(id=function.id)
```

## Call the Function with Sample Data


```python
# Call with sample DMS instances
call = client.functions.call(
    external_id="print-data-sample",
    data={
        "instances": {
            "items": [
                {
                    "externalId": "asset-001",
                    "space": "my_space",
                    "instanceType": "node",
                    "type": {
                        "space": "model_space",
                        "externalId": "Asset",
                        "version": "v1"
                    },
                    "properties": {
                        "model_space": {
                            "AssetView/v1": {
                                "name": "Production Asset",
                                "type": "PUMP_STATION"
                            }
                        }
                    }
                }
            ]
        }
    }
)

print(f"Function called: {call.id}")
print(f"Status: {call.status}")

# Wait for completion
print("\\nWaiting for completion...")
call.wait(timeout=60)

print(f"\\nStatus: {call.status}")
print(f"\\nResponse:")
print(json.dumps(call.response, indent=2))
```

## View Function Logs

The actual data structure is printed to the logs.


```python
# Get logs from the specific call
logs = client.functions.calls.get_logs(call_id=call.id)

print("FUNCTION LOGS:")
print("=" * 80)
for log_entry in logs:
    print(log_entry.message)
```
